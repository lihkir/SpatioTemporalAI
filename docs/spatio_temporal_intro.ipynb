# Modelos Espacio-Temporales

````{admonition} Introducción
:class: tip

- Durante décadas, el análisis de fenómenos que varían en el espacio y en el tiempo (calidad del aire, clima, tráfico, epidemias, recursos energéticos) dependía de modelos que trataban la dimensión ***espacial*** y la dimensión ***temporal*** de forma separada, o que asumían independencia entre observaciones cercanas. Los ***Modelos Espacio-Temporales*** cambian radicalmente este paradigma: en lugar de ignorar la estructura de dependencia entre ubicaciones y momentos en el tiempo, se construyen ***arquitecturas y modelos estadísticos*** capaces de ***capturar simultáneamente la correlación espacial y la dinámica temporal*** directamente a partir de datos georreferenciados y series de tiempo multivariadas.

- Cada componente de un modelo espacio-temporal captura una parte distinta del fenómeno: la ***dependencia espacial*** entre estaciones o ubicaciones cercanas, y la ***dependencia temporal*** entre observaciones consecutivas. Esta capacidad de modelar conjuntamente ambas dimensiones, combinada con la disponibilidad de grandes volúmenes de datos geoespaciales y de cómputo acelerado por GPU, es la razón por la cual los modelos espacio-temporales se han convertido en una herramienta clave para la predicción y el monitoreo ambiental, energético y de salud pública en la última década.
````

## De los modelos clásicos al modelado espacio-temporal

- En el curso de aprendizaje supervisado trabajamos con modelos que reciben un conjunto ***fijo*** de características diseñadas explícitamente, asumiendo generalmente que las observaciones son independientes entre sí. Estos modelos son interpretables y eficientes, pero ignoran la ***autocorrelación espacial*** (observaciones cercanas geográficamente tienden a parecerse) y la ***autocorrelación temporal*** (observaciones cercanas en el tiempo tienden a parecerse), limitando su desempeño cuando los datos tienen estructura de red o de campo aleatorio.

- Cuando los datos provienen de ***múltiples ubicaciones monitoreadas a lo largo del tiempo*** (estaciones de calidad del aire, sensores meteorológicos, nodos de tráfico, casos epidemiológicos por región), modelar cada serie de forma aislada se vuelve ineficiente e inexacto. El modelado espacio-temporal resuelve este problema incorporando ***estructuras de dependencia espacial (kriging, procesos gaussianos, grafos) y temporal (recurrencia, atención, procesos autorregresivos)***, optimizando de forma conjunta la representación espacial y la dinámica temporal.

````{admonition} Observación
:class: important

- El modelado espacio-temporal ***no reemplaza*** los fundamentos estadísticos y de optimización vistos previamente; los ***extiende***. Conceptos como sobreajuste, generalización, validación cruzada y selección de modelos siguen siendo centrales, pero ahora se aplican considerando ***dependencias entre observaciones vecinas en el espacio y en el tiempo***, lo que exige un manejo más cuidadoso de la validación cruzada espacial, la regularización y la elección de estructuras de covarianza o de vecindad.
````

## Aplicaciones que veremos a lo largo del curso

- El poder de los Modelos Espacio-Temporales se aprecia mejor a través de sus aplicaciones. A lo largo del curso trabajaremos, entre otros, con los siguientes casos de uso:

    - ***Medio ambiente***: predicción de la calidad del aire (PM2.5, PM10, ozono) a partir de redes de estaciones de monitoreo y datos satelitales.
    - ***Energías renovables***: pronóstico de velocidad del viento y radiación solar considerando la correlación entre estaciones cercanas.
    - ***Salud pública***: modelado de la propagación espacio-temporal de enfermedades y epidemias entre regiones.
    - ***Clima e hidrología***: interpolación y pronóstico de variables climáticas (temperatura, precipitación) sobre campos espaciales continuos.
    - ***Transporte y movilidad***: predicción de flujos de tráfico y patrones de movilidad urbana sobre redes viales.
    - ***Aprendizaje profundo espacio-temporal***: redes convolucionales recurrentes (ConvLSTM), grafos neuronales (GNN) y arquitecturas basadas en atención para datos espacio-temporales.

- Cada uno de estos bloques combina ***fundamento teórico riguroso*** (procesos gaussianos, kriging, modelos autorregresivos espaciales, derivaciones y algoritmos) con ***implementación práctica*** en `Python`, usando `TensorFlow`/`Keras` y librerías especializadas en análisis geoespacial, y en algunos casos `PySpark` para escenarios de gran volumen de datos.
